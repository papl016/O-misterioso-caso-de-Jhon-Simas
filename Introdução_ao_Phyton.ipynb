## AULA 2

Contexto:

Um app de estástiscas de futebol está avaliando o desempenho de um jogador com base em diferentes critérios, atribuindo pontos por cada um deles:

Gol marcado: 10 pontos

Assistência: 8 pontos

Jogo sem levar cartão: 6 pontos


PROBLEMA:

Júnior fez um gol, deu uma assistência e não levou cartão em um jogo. Quantos pontos ele acumulou no jogo?

```python
gol_marcado = 10
assistência = 8
jogo_sem_levar_cartão = 6

pontos_rodrigo = gol_marcado + assistência + jogo_sem_levar_cartão

print(f"Os pontos totais do Rodrigo são: {pontos_rodrigo}")
```
